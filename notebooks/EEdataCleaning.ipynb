```python
import pandas as pd
import glob
import os

from pandas.conftest import datetime_series
```

```python
folder = r"Data/EarthEngine_Exports"

files = glob.glob(os.path.join(folder, "*.csv"))
```

```python
dfs = []

for file in files:
    df = pd.read_csv(file)
    dfs.append(df)
```

```python
dataset = pd.concat(dfs, ignore_index=True)
dataset.shape
```

```python
dataset.drop(columns=[
    "system:index",
    ".geo"
], inplace=True)
```

```python
dataset.duplicated().sum()
```

```python
dataset.shape
```

```python
duplicates = dataset[dataset.duplicated(keep=False)]
duplicates.head(20)
```

```python
duplicates[duplicates['BuiltSurface']==3994].head()
```

```python
duplicates.groupby(list(dataset.columns)).size().sort_values(ascending=False).shape
```

```python
dataset = dataset.drop_duplicates()
```

```python
dataset.shape
```

```python
dataset.duplicated().sum()
```

```python
import numpy as np

dataset["WindSpeed"] = np.sqrt(
    dataset["UWind"]**2 +
    dataset["VWind"]**2
)
```

```python
dataset.drop(
    columns=["UWind","VWind"],
    inplace=True
)
```

```python
dataset.head()
```

```python
dataset.shape
```

```python
dataset["Humidity"] = 100 * np.exp(
    (17.625*dataset["DewPoint"])/(243.04+dataset["DewPoint"])
    -
    (17.625*dataset["AirTemp"])/(243.04+dataset["AirTemp"])
)
```

```python
dataset.drop(columns=["DewPoint"], inplace=True)
```

```python
dataset.to_csv(
    "Data/Delhi_Master_Dataset.csv",
    index=False
)
```
